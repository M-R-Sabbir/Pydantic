```python
from pydantic import BaseModel

class Patient(BaseModel):
    name:str
    age:int
# making pydantic class

Patient_info ={"name":"Sabbir","age":"24"}
Patient1 = Patient(**Patient_info)         # dictionary unpacking
# if the validation phase is passed we will use instantiated object in further funcs


users = []

def user_data_insertion(patient:Patient):      # passing patient object and telling this object dtype is Patient
    users.append((patient.name,patient.age))
    print(users)
    print("Inserted Successfully")

user_data_insertion(Patient1)


# This is how pydantic solves type validation .
# see it also accept if we give "24" so can handle this also,smart enough to convert this "24" to 24
```

Now let's use it for bigger classes

```python
from pydantic import BaseModel
from typing import List,Dict,Optional


class Patient(BaseModel):
    name : str
    age : int
    weight : float
    married : bool=False
    things : Optional[list]= None
    allergies : List[str]
    contact : Dict[str,str]

patient_info = {'name':"Sabbir","age":24,"weight":70.9,"allergies":["Hello","world"],
                'contact':{"mail":"abc@gmail","call":"1627113136"},"married":True}

patient2 = Patient(**patient_info)

def insertion(patient:Patient):
    print(patient.name)
    print(patient.age)
    print(patient.things)

insertion(patient2)
```

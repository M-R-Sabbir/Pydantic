## model_validation
If we want to validate multiple data combined, we use model_validation

```python
from pydantic import BaseModel,EmailStr,AnyUrl,model_validator
from typing import List,Dict,Optional


class Patient(BaseModel):
    name : str
    age : int
    email: EmailStr              
    linkdin:AnyUrl                 
    weight : float
    married : bool=False
    things : Optional[list]= None
    allergies : List[str]
    contact : Dict[str,str]

    @model_validator(mode='after')
    def validate_emergency_contact(cls,model):
        if model.age>60 and 'emergency' not in model.contact:
            raise ValueError("60+ aged people must have emergency contact")
        else:
            return model



patient_info = {'name':"Sabbir","age":64,"weight":70.9,"allergies":["Hello","world"],
                'contact':{"mail":"abc@gmail","call":"1627113136",'emergency':"91219935"},
                "married":True,"email":"abc@gmail.com","linkdin":"https://linkdin.com/123"}


patient2 = Patient(**patient_info)

def insertion(patient:Patient):
    print(patient.name)
    print(patient.age)
    print(patient.things)
    print(patient.email)
    print(patient.linkdin)


insertion(patient2)
```

The thing is we wanna work with more than 1 variable, or whole model, 

like we want if any consumer is 60+ aged ,he should have an emergency contact.

so we have to make sure both things using model validation
